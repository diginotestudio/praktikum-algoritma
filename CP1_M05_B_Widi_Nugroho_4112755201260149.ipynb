Konsep relasional

```python
score = 15
print(score > 10)
print(score < 20)
print(score >= 15)
print(score == 15)
print(score != 15)
```

STEP 1  AND

```python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

AND menghasilkan TRUE jika 2 atau lebih pernyataan bernilai TRUE. Jika salah satu atau lebih dari penyataan-penyataan itu bernilai FALSE, maka hasil AND akan FALSE

STEP 2  OR

```python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

OR menghasilkan FALSE ketika semua pernyataan bernilai FALSE. Jika salah satu pernyatan bernilai TRUE, maka hasil dari OR akan menjadi TRUE.

STEP 3  NOT

```python
print(not True)
print(not False)
```

not True = False | not False = True

STEP 4  REAL CASE

```python
score = 15
print(score > 10 and score < 20)

score = 25
print(score > 10 and score < 20)
```

output berubah ketika score menjadi 25 karena logika yang dipakai adalah logika AND dan nilai 15 memenuhi syarat dari kedua pertanyaan ( 15 > 10 == TRUE | 15 < 20 == TRUE) sehingga hasilnya TRUE. sedangkan pada nilai 25, hasilnya FALSE karena tidak memenuhi kedua syarat (25 > 10 == TRUE | 25 < 20 == FALSE)

Activity 6  Debugging Challenge

Case 1  TypeError

```python
num = 10
print("Hasil: " + num)
```

Jenis data "Hasil: " = **String** | Jenis data num = **Integer**

Error terjadi karena tipe data String tidak bisa digabungkan dengan tipe data integer.

```python
num = 10
print("Hasil: " + str(num))
```

str(num) menyelesaikan masalah karena code ini berfungsi untuk mengubah tipe data integer num  menjadi tipe data String, dan karena string bisa digabung dengan string, maka tidak muncul error.

Case 2  -  ZeroDivisionError

```python
x = 5
y = 0
print(x / y)
```

Jenis Error yang muncul adalah "dibagi dengan nol". Hal ini menyebabkan error karena dalam operasi hitung matematika, semua nilai yang dibagi dengan nol, hasilnya tidak terdefinisikan, sehingga menimbulkan error.

Perbaikan:

```python
x = 5
y = 0
if y != 0:
  print(x / y)
else:
  print("Pembagian tidak dapat dilakukan karena y = 0")
```

Setelah diperbaiki dengan IF, tidak muncul error.

Mengapa kondisi y != 0 dapat mencegah ZeroDivisionError karena fungsi if sudah melakukan filter terhadap operasi aritmatika di atas. Percabangan IF pada syntax dapat berarti, jika nilai y tidak sama dengan 0, maka lakukan perhitungan x / y, tapi jika tidak (y bernilai 0) maka cetak output "Pembagian tidak dapat dilakukan karena y = 0". Dan karena y = 0, maka hasil code itu adalah output cetak "Pembagian tidak dapat dilakukan karena y = 0".

Activity 7  Mini Challenge  Score Checker

Sebuah sistem menghitung nilai akhir dari nilai awal dan bonus. Program juga harus memeriksa apakah nilai akhir memenuhi batas kelulusan dan berada dalam rentang nilai valid.

1. Buat variabel score = 75.
2. Buat variabel bonus = 10.
3. Hitung final_score = score + bonus.
4. Tampilkan score awal, bonus, dan final_score.
5. Tampilkan hasil pengecekan apakah final_score minimal 70.
6. Tampilkan hasil pengecekan apakah final_score berada pada rentang 70 sampai 100.

```python
score = 75
bonus = 10
final_score = score + bonus

print ("Score : " + str(score))
print ("Bonus : " + str(bonus))
print ("Final Score : " + str(final_score))
print (final_score >= 70)
print (final_score >= 70 & final_score <= 100)
```
